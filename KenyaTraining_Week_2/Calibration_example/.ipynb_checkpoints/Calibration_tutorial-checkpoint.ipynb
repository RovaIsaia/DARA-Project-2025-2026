# 0. Data first look

FITS is a special data format to store astronomical data, which stands for Flexible Image Transport System. A FITS file contains one or many HDUs (Header + Data Unions). For an image fits file, the header contains information of the observation. The data union contains a 2D array with information about the number of photon counts (i.e. brightness) received by each pixel. Each value in the 2D array represents the brightness received by that area of the detector.

Before we start to use the data, let's have a quick look of how the data is like.

In the following steps, we'll open a raw image, read the observation infomation, and print the image on the screen.

```python
# Import a python package to read fits file.
from astropy.io import fits

# Read the hdu (Header + Data Unions) of the image file.
filename = data_path + 'raw.fits'
hdulist = fits.open(filename)
hdulist.info()

# As you will see, our image data has one PRIMARY HDU, which contains a 2080 * 4128 integer array.
```

```python
# Header contains critical information of the observation, which was either written by the observer or generated by a pipeline.
prihdr = hdulist[0].header
prihdr
```

```python
# Quiz: What telescope took this data? When was the data taken?
# Try these commands:
print('The observation date is:', prihdr['DATE-OBS'])
print('The observation time is:', prihdr['UT'])
print('The exposure time is:', prihdr['EXPTIME'], 'seconds')
```

After having a general idea of the observation information, let us see how the raw image looks like.

```python
# An image is basically a two dimensional array: Numbers in the array represents the amplitude of signal at a certain (x, y) coordinate.
# Read the image data array:
img_data = hdulist[0].data

# Check the pixel number:
print(img_data.shape)
```

```python
# You can even see the numbers in the array!
print(img_data)
```

```python
# If we use a color-value projection to plot the array above, we get an image showing the flux distribution.
import matplotlib.pyplot as plt

# Show the image
plt.imshow(img_data, cmap='gray',norm=matplotlib.colors.LogNorm())
plt.colorbar();
```

```python
# Remember to close the HDU after use!
hdulist.close()
```

# Calibration

A raw astronomical image is not only the pure sky signal. It contains several other components from instruments, i.e.,

$\rm{Raw Image = Bias + Dark\,Current + Flat * Sky\,Signal}$

Or in other words, our goal is to derive the sky signal through:

$\rm{Sky\,Signal = (Raw Image - Bias - Dark\,Current)/Flat}$

The values in the HDU array can be manipulated to remove different types of noise as it is already digitalized. For example, if you know there is an added-on noise, you can simply subtract that value from each element in the array to obtain a cleaner image. We use different methods to reduce noise from three common sources:

# 1. Bias subtraction

Bias is an offset voltage (which translates into some non-zero number of counts) added to every pixel in the image to ensure that when voltages are converted to counts there is never a negative count.
The bias value is roughly the same across the CCD chip, though it’s not uncommon to have “bad” columns and pixels in which the bias level is consistently offset from the rest of the chip.
The bad columns in a CCD are typically stable over a very long time.

Note that the bias doesn’t depend on exposure time. That’s because a bias exposure is a zero-second exposure in which the camera simply reads the chip out. To remove the bias, we simply subtract it from the raw image.

```python
# Read the raw image.
fits_filename = data_path + 'raw.fits'
raw_hdulist = fits.open(fits_filename)
raw = raw_hdulist[0].data.astype('float')
```

```python
# Read the bias image.
fits_filename = data_path + 'bias.fits'
hdulist = fits.open(fits_filename)
bias = hdulist[0].data.astype('float')

# Plot the bias image to have a first look.
plt.imshow(bias, cmap='gray', norm=matplotlib.colors.LogNorm())
plt.colorbar();
```

```python
# Subtract the bias from the raw image.
bias_subtracted = raw - bias

# Plot the bias-subtracted image
plt.imshow(bias_subtracted, cmap='gray', norm=matplotlib.colors.LogNorm())
plt.colorbar();
```

# 2. Dark subtraction

Dark current is defined as the current flowing through the detector when there is no incident photon flux. It determines the base current level that photocurrent must exceed in order to be detected. It is a base of output that should be subtracted.

Dark current depends on the temperature of the sensor, and thus will change during the night and also with how long the exposure time is. A dark frame (or dark image) is an image taken with the camera shutter closed and with the same exposure time as your science (raw) image.

```python
# Read the dark image
fits_filename = data_path + 'dark.fits'
hdulist = fits.open(fits_filename)
dark = hdulist[0].data.astype('float')
```

```python
# Plot the dark image
plt.imshow(dark, cmap='gray', norm=matplotlib.colors.LogNorm())
plt.colorbar();
```

```python
# Compare the dark exposure time with observation time
dark_time = float(hdulist[0].header['EXPTIME'])
obs_time = float(raw_hdulist[0].header['EXPTIME'])
ratio = obs_time / dark_time
print(f'Observation exposure time is {obs_time:.5f} seconds.')
print(f'Dark exposure time is {dark_time:.5f} seconds.')
print(f'Ratio of obs time vs. dark exp time is {ratio:.4f}.')
```

```python
# Subtract dark current from the image
dark_subtracted = bias_subtracted - dark #* ratio

# plot the dark subtracted image
#cm = matplotlib.cm.gray
#cm.set_bad('black')
plt.imshow(dark_subtracted, cmap='gray', norm=matplotlib.colors.LogNorm())
plt.colorbar();
```

# 3. Flat correction

Different parts of a CCD respond to signal differently. You might have noticed such a thing in daily photography -- vignetting appears at the corners of a phone photo. In CCD data reduction, a photo of a uniform light source (usually twilight sky or dome flat lamp) helps the observer to understand how the sensitivity changes among the CCD. Usually on a flat image, you'll see not only the vignetting, but also shadows of dust grains in the optical path, etc.

Since the sensitivity is nonuniform, we need to apply correction factors to every local pixel, to make them even. For example, an original signal of 100 (photon count, Jansky, Watt/square meter...) might be reduced to 20 at A pixel and 40 at B pixel. To make them back to the same level, we can calculate 20/1 and 40/2, or 20/0.2 and 40/0.4, etc (i.e., apparent signal strength divided by a sensitivity factor). How do we know this sensitivity factor in practice? It is usually derived from normalisation among the whole flat image, as will be explained in the followings.

```python
# Let's first see how the flat image looks like! Here we open a flat image:
fits_filename = data_path + 'flat.fits'
hdulist = fits.open(fits_filename)
flat = hdulist[0].data.astype('float')

# And plot the flat image:
plt.imshow(flat, cmap='gray', norm=matplotlib.colors.LogNorm())
plt.colorbar();
```

The flux received at a pixel is equivalent to its sensitivity since the light source for flat is uniform. For instance, if there are 10 photons at A pixel and 20 photons at B pixel, B pixel will have twice the sensitivity of A pixel. Note that the sensitivity here is relative, so we can multiply or divide the flat by the same scaling factor without changing its nature.

Usually, we scale the flat with its median value. There are two reasons using the median: First, most flat values will be around 1 after rescaling, and thus the flat correction. Second, median is better than mean as it is not biased by a few extreme values (e.g., cosmic ray) on the flat image.

```python
# Normalizing the flat by rescaling with respect to the median
flat_normed = flat / np.median(flat)

# Remember that the "flat_normed" (sensitivity factor) will be in the denominator.
# It is useful to check zero value first to avoid warnings caused by dividing by zero.
if np.argwhere(flat_normed==0).any(): # if have zeros
    print('There is zero in the flat.')
```

```python
# Replace the zeros in the flat by a small number
flat_normed[flat_normed==0] = 1e-5
```

```python
# Divide the apparent image by sensitivity factor (normalised flat)
flat_divided = dark_subtracted / flat_normed

# Plot the final image
plt.imshow(flat_divided, cmap='gray', norm=matplotlib.colors.LogNorm())
plt.colorbar();
```

This is the final product of sky image, with all instrumental effects removed. Now it consists of three components:

$\rm{Sky\,Signal = Source + Sky\,Background + Noise}$

Source is the real signal that we want. Sky background is the emission from the sky. And noise is what is left from the steps above (you can never get rid of noise, it can only be reduced!)
